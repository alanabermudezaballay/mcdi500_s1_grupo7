## 5. Estimación puntual e intervalos de confianza

```python
!pip install scipy
```

```python
import numpy as np
import scipy.stats as stats

data = df['MaxTemp'].dropna()

n = len(data)
media = data.mean()
std = data.std()

intervalo = stats.t.interval(0.95, n-1, loc=media, scale=std/np.sqrt(n))

media, intervalo
```

El análisis de estimación puntual indica que la temperatura máxima promedio se sitúa aproximadamente en torno a los **23.22°C**.

Asimismo, el intervalo de confianza del 95% permite establecer que el valor real de la temperatura máxima promedio se encuentra entre **23.18°C y 23.25°C**.

Esto implica que existe un alto nivel de confianza en que el parámetro poblacional se ubica dentro de dicho rango, reduciendo la incertidumbre en la estimación y proporcionando una base sólida para el análisis estadístico.

## 6. Prueba de hipótesis

```python
t_stat, p_value = stats.ttest_1samp(data, 20)

t_stat, p_value
```

Con el propósito de evaluar si la temperatura máxima promedio es superior a 20°C, se plantearon las siguientes hipótesis:

- H0: la media es menor o igual a 20°C  
- H1: la media es mayor a 20°C  

A partir de la prueba t de una muestra, se obtuvo un valor p menor a 0.001.

Dado que este valor es menor a 0.05, se rechaza la hipótesis nula, concluyendo que existe evidencia estadística suficiente para afirmar que la temperatura máxima promedio es superior a 20°C.

Este resultado es consistente con el análisis descriptivo y el intervalo de confianza previamente calculado.
