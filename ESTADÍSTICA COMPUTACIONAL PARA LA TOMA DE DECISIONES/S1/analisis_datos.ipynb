# Preparación y Limpieza de Datos 
*Carga, limpieza y validación del dataset.*



## 1. Carga del Dataset
En esta etapa, importamos la librería *pandas* para procesar los datos estructurados.

```python
import pandas as pd
df = pd.read_csv('weatherAUS.csv') # Asegúrate de que el nombre sea exacto
df.head() # Esto te mostrará las primeras filas de tu dataset
```

## 2. Revisión de tipos de variables
Analizamos el formato de cada columna para asegurar que los números sean tratados como números y las fechas como fechas.

```python
# Ver el tipo de dato de cada columna
df.dtypes
```

## 3. Identificación de valores faltantes y duplicados
Analizamos la calidad de los datos detectando celdas vacías (valores nulos) y posibles filas duplicadas que puedan sesgar el análisis.

```python
# Contar valores nulos (NaN) por columna
print("Valores nulos por columna:")
print(df.isnull().sum())

# Contar filas duplicadas
print("\nCantidad de filas duplicadas:", df.duplicated().sum())
```

## 4. Limpieza de datos
En este paso, procedemos a eliminar registros duplicados para asegurar la integridad de la muestra y tratamos los valores nulos para evitar errores en los cálculos estadísticos.

```python
# 1. Eliminar filas duplicadas
df.drop_duplicates(inplace=True)

# 2. Rellenar nulos de columnas numéricas con la mediana
# Esto mantiene la integridad de los datos sin perder filas
for col in df.select_dtypes(include=['float64']).columns:
    df[col] = df[col].fillna(df[col].median())

print("Limpieza completada: Se eliminaron duplicados y se imputaron valores faltantes.")
```

## 5. Descripción inicial del dataset
Tras la limpieza, generamos un resumen estadístico que describe las principales métricas del dataset, permitiéndonos entender la distribución de las variables meteorológicas.

```python
# Generar resumen estadístico de las variables numéricas
df.describe()
```

Interpretación del Resumen Estadístico:

count: Indica que contamos con 145,460 registros válidos.

mean: Es el promedio de cada variable meteorológica.

std: (Desviación estándar) Mide la dispersión de los datos respecto al promedio.

min/max: Muestran los valores extremos registrados.

Cuartiles (25%, 50%, 75%): Dividen la distribución, permitiéndonos identificar el comportamiento central de las variables.

```python
# Verificación final
print(f"Dimensiones finales del dataset: {df.shape}")
print("¿Quedan valores nulos?:", df.isnull().sum().sum() > 0)
```

```python
# Verifica cuántos nulos hay por columna
nulos_por_columna = df.isnull().sum()
print(nulos_por_columna)
```

Se realizó un conteo de valores faltantes (NaN) por cada variable. Los resultados muestran la presencia de celdas sin información, lo cual es una característica esperada en datasets meteorológicos de gran escala. Estos nulos han sido identificados para su gestión estratégica en las etapas posteriores del proyecto.

```python
# Limpieza de variables categóricas
cat_cols = ['WindGustDir', 'WindDir9am', 'WindDir3pm', 'RainToday', 'RainTomorrow']

for col in cat_cols:
    # Rellenamos los nulos con la moda (el valor más común)
    df[col] = df[col].fillna(df[col].mode()[0])

print("Limpieza de variables categóricas completada.")
```

Se identificó que las variables categóricas presentaban valores nulos que no fueron abordados por la imputación numérica basada en la mediana. Se aplicó una imputación basada en la moda (valor más frecuente) para estas variables, asegurando así la completitud del dataset sin alterar la distribución de las categorías, permitiendo que el conjunto de datos esté íntegramente disponible para las siguientes etapas de modelamiento predictivo.

```python
# Limpieza forzada y profesional de columnas de dirección del viento
cols_direccion = ['WindGustDir', 'WindDir9am', 'WindDir3pm']

for col in cols_direccion:
    # Calculamos la moda
    moda = df[col].mode()[0]
    # Usamos loc para asegurar que estamos modificando el DataFrame original sin ambigüedades
    df.loc[:, col] = df[col].fillna(moda)

# Verificación final
print("Limpieza finalizada con éxito. Nulos restantes en direcciones:", df[cols_direccion].isnull().sum().sum())
```

## Análisis Exploratorio de Datos

En esta sección se realiza un análisis exploratorio del dataset limpio,
con el objetivo de entender el comportamiento de las variables
y detectar patrones relevantes.

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv('weatherAUS_limpio.csv')
```

```python
variables = ['MinTemp', 'MaxTemp', 'Rainfall', 'Humidity3pm']
df[variables].describe()
```

```python
variables = ['MinTemp', 'MaxTemp', 'Rainfall', 'Humidity3pm']
df[variables].describe()
```

```python
plt.figure(figsize=(6,4))
sns.histplot(df['MaxTemp'], bins=30)
plt.title('Distribución de la temperatura máxima')
plt.show()
```

### Interpretación de la temperatura

La temperatura máxima presenta una distribución cercana a la normal,
lo que indica que los valores se concentran en un rango promedio.

```python
plt.figure(figsize=(6,4))
sns.histplot(df['Rainfall'], bins=30)
plt.title('Distribución de la precipitación')
plt.show()
```

### Interpretación de la precipitación

La precipitación presenta una fuerte asimetría,
ya que la mayoría de los días no presentan lluvia,
pero existen valores extremos que elevan el promedio.

Estos resultados son relevantes en el contexto de la predicción climática,
ya que permiten identificar patrones que pueden influir en la ocurrencia de precipitaciones,
lo cual es importante para la toma de decisiones bajo incertidumbre.

```python
plt.figure(figsize=(6,4))
sns.histplot(df['Humidity3pm'], bins=30)
plt.title('Distribución de la humedad')
plt.show()
```

### Interpretación de la humedad

La humedad presenta una alta variabilidad,
lo que refleja cambios importantes en las condiciones climáticas.

Se observa una correlación positiva entre la temperatura mínima y máxima,
lo cual es esperable debido a su relación natural.

Por otro lado, la humedad presenta una relación inversa con la temperatura máxima,
lo que sugiere que días más cálidos tienden a ser menos húmedos.

```python
plt.figure(figsize=(6,4))
sns.boxplot(x=df['MaxTemp'])
plt.title('Boxplot temperatura máxima')
plt.show()
```

### Interpretación del boxplot

El gráfico permite identificar la presencia de valores atípicos,
lo que indica condiciones poco frecuentes dentro del comportamiento climático.

```python
variables = ['MinTemp', 'MaxTemp', 'Rainfall', 'Humidity3pm']

plt.figure(figsize=(6,5))
sns.heatmap(df[variables].corr(), annot=True, cmap='coolwarm')
plt.title('Matriz de correlación')
plt.show()
```

### Interpretación de la correlación

Se observa una relación positiva entre la temperatura mínima y máxima,
lo cual es coherente con el comportamiento climático.

Por otro lado, la precipitación presenta correlaciones más débiles,
lo que sugiere que depende de múltiples factores.

Estos resultados son relevantes en la toma de decisiones bajo incertidumbre.
